# Simple Precipitation Simulator

A linearly-advected Gaussian random field on a periodic domain. A model of precipitation is modelled by thresholding it using a moving envelope.

- **Output variables**: `accumulation`, `rain` (instantaneous rate), `phi` (value of the gaussian random field), `envelope` 

## Model

The Gaussian field $\phi$, which has zero mean and covariance given by:

$$
\operatorname{Cov}\big(\phi(\mathbf{x}, t),\, \phi(\mathbf{x} + \mathbf{r}, t + h)\big)
= s^2 \exp\left(-\frac{|\mathbf{r} - \mathbf{c}h|^2}{2\ell^2} - \frac{h^2}{2\tau_g^2}\right)
$$

This is thresholded using a moving Gaussian envelope:

$$
R = \max(\phi - \theta,\, 0), \qquad
\theta(\mathbf{x}, t) = \theta_0 - \Delta\theta \exp\left(-\frac{|\mathbf{x} - \mathbf{x}_0 - \mathbf{c}_e t|^2}{2L_{env}^2}\right).
$$

The outputs at $t_n = n\,\Delta t_{out}$ are

$$
A_n = \int_{t_{n-1}}^{t_n} R \, dt \qquad \text{and} \qquad R(t_n).
$$

| code | `s` | `ell` | `tau_g` | `cx`, `cy` | `cex`, `cey` | `x0`, `y0` | `L_env` | `theta0` | `dtheta` |
|---|---|---|---|---|---|---|---|---|---|
| symbol | $s$ | $\ell$ | $\tau_g$ | $\mathbf{c}$ | $\mathbf{c}_e$ | $\mathbf{x}_0$ | $L_{env}$ | $\theta_0$ | $\Delta\theta$ |

```python
from IPython.display import HTML

from autosim.experimental.simulations import SimplePrecipitation
from autosim.utils import plot_spatiotemporal_video
```

```python
params = {
    "s": 1.0,
    "ell": 3.5,
    "tau_g": 8.0,
    "cx": 0.8,
    "cy": 0.8,
    "cex": 1.0,
    "cey": 1.0,
    "x0": 20.0,
    "y0": 20.0,
    "L_env": 25.0,
    "theta0": 4.0,
    "dtheta": 4.0,
}

sim = SimplePrecipitation(
    parameters_range={k: (v, v) for k, v in params.items()},
    log_level="progress_bar",
    T=400.0,
    dt_out=20.0,
)

batch = sim.forward_samples_spatiotemporal(n=2, random_seed=42)

print("data shape:", batch["data"].shape, "[member, time, x, y, channels={accumulation, rain, phi, envelope}]")
```

```python
anim = plot_spatiotemporal_video(
    batch["data"],
    batch_idx=0,
    channel_names=["accumulation", "rain", "phi", "envelope"],
    preserve_aspect=True,
)
HTML(anim.to_jshtml())
```
